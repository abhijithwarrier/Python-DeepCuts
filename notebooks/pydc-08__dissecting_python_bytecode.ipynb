**PYTHON SCRIPT TO _UNDERSTAND HOW PYTHON TRANSLATES CODE INTO BYTECODE USING THE `dis` MODULE_. 🐍⚙️**

Python does not directly execute your source code.
CPython **compiles** it into a sequence of bytecode instructions — low-level operations executed by the Python Virtual Machine (PVM).

In this DeepCut, we explore:

- How Python compiles code to bytecode
- How to inspect bytecode using `dis.dis()`
- What stack-based execution looks like
- How control flow (loops, conditionals) appears in bytecode
- Performance insights you can gain from bytecode

---

## 📦 Import Standard Library

```python
import dis
```

---

## 🧩 Snippet 1 — Python compiles functions into bytecode instructions

Let's examine a simple function and see how Python translates it.

```python
def add(a, b):
    return a + b

dis.dis(add)
```

---

## 🔍 Snippet 2 — How if-statements appear in bytecode

Conditional logic becomes jumps in bytecode.

```python
def check(x):
    if x > 10:
        return "big"
    return "small"

dis.dis(check)
```

---

## 🔄 Snippet 3 — Loops are turned into jump instructions and iteration opcodes

Python uses a stack machine; loops push and pop values via opcodes.

```python
def loop(n):
    total = 0
    for i in range(n):
        total += i
    return total

dis.dis(loop)
```

---

## 🧠 Snippet 4 — Bytecode operates on a stack, not registers

Each opcode pushes or pops values.

Example: a = (x + y) * z

```python
def expr(x, y, z):
    return (x + y) * z

dis.dis(expr)
```

---

## 🧬 Snippet 5 — Functions carry their own code object

We can inspect:
- constants
- variable names
- stack size
- instructions

```python
def sample(a, b):
    c = a * b
    return c + 10

code = sample.__code__

print("Co varnames:", code.co_varnames)
print("Co names:", code.co_names)
print("Co constants:", code.co_consts)
print("Stack size:", code.co_stacksize)
dis.dis(sample)
```

---

## 🔧 Snippet 6 — dis.Bytecode gives a structured view of each opcode

Useful for static analysis tools, linters, and compilation research.

```python
bytecode = dis.Bytecode(sample)

for instr in bytecode:
    print(instr.opname, "→", instr.argval)
```

---

## ✅ One-liner Takeaway

**Python compiles your code into stack-based bytecode — and `dis` lets you peek directly into the instructions the CPython VM executes.**
