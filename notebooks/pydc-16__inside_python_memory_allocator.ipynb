**PYTHON SCRIPT TO _UNDERSTAND HOW PYTHON ALLOCATES, REUSES, AND MANAGES MEMORY INTERNALLY_. 🐍🧠**

Python does not rely entirely on the system allocator (`malloc`).
Instead, CPython uses its own layered memory allocator optimized for:
- small objects
- fast allocation
- reduced fragmentation

In this DeepCut, we explore:
- pymalloc’s design
- arenas, pools, and blocks
- why memory is not always returned to the OS
- practical implications for real applications

---

## 📦 Import Standard Library

```python
import sys
import tracemalloc
```

---

## 🧩 Snippet 1 — All Python objects are heap-allocated

Even small integers and strings live on the heap.

```python
a = 10
b = "hello"
c = [1, 2, 3]

print(type(a), type(b), type(c))
```

---

## 🧠 Snippet 2 — pymalloc handles small objects efficiently

Objects smaller than ~512 bytes use Python’s internal allocator.

```python
sizes = [sys.getsizeof(i) for i in range(10)]
print(sizes)
```

---

## 🔄 Snippet 3 — How pymalloc structures memory

pymalloc uses a hierarchy:
- Arenas (large chunks from OS)
- Pools (fixed-size blocks)
- Blocks (individual object slots)

---

## 🧠 Snippet 4 — Python keeps memory for reuse

Freeing objects does not always return memory to the OS.

```python
lst = [i for i in range(1_000_000)]
del lst

print("List deleted")
```

---

## 🔍 Snippet 5 — Inspect Python memory allocations

tracemalloc tracks allocations at the Python level.

```python
tracemalloc.start()

data = [str(i) for i in range(10000)]

current, peak = tracemalloc.get_traced_memory()
print(f"Current: {current}, Peak: {peak}")

tracemalloc.stop()
```

---

## 🧬 Snippet 6 — Design trade-offs in pymalloc

pymalloc favors:
- speed over memory return
- predictable allocation
- reduced OS calls

---

## 🧠 Snippet 7 — What this means in real systems

Understanding pymalloc explains:
- why memory “leaks” aren’t always leaks
- why restarting processes frees memory
- why object pooling works well

---

## ✅ One-liner Takeaway

**CPython uses pymalloc to aggressively reuse memory — optimizing speed and stability over returning memory to the OS.**
