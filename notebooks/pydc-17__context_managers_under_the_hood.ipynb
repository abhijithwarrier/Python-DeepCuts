The `with` statement looks simple — but it triggers a precise protocol
involving `__enter__` and `__exit__`.

---

## 📦 Import Standard Library

```python
from contextlib import contextmanager
```

---

## 🧩 Snippet 1 — The __enter__ and __exit__ protocol

Any object with:
- __enter__()
- __exit__()

Can be used in a `with` block.

```python
class DemoContext:
    def __enter__(self):
        print("Entering context")
        return "Resource"

    def __exit__(self, exc_type, exc_value, traceback):
        print("Exiting context")
        print("Exception:", exc_type)
        return False  # Do not suppress exceptions

with DemoContext() as value:
    print("Inside with:", value)
```

---

## 🧠 Snippet 2 — Exception handling inside __exit__

If __exit__ returns True, the exception is suppressed.

```python
class SuppressContext:
    def __enter__(self):
        return None

    def __exit__(self, exc_type, exc_value, traceback):
        print("Handled exception:", exc_type)
        return True  # Suppress error

with SuppressContext():
    1 / 0  # No crash
```

## 🔄 Snippet 3 — Equivalent try/finally expansion

The `with` statement expands roughly into:

manager = EXPR
value = manager.__enter__()
try:
    BLOCK
finally:
    manager.__exit__()

---

## 🧬 Snippet 4 — Generator-based context managers

contextlib allows building managers with a generator.

```python
@contextmanager
def simple_context():
    print("Setup")
    yield "Resource"
    print("Cleanup")

with simple_context() as value:
    print("Using:", value)
```
