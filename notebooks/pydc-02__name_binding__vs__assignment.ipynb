**PYTHON SCRIPT TO _UNDERSTAND HOW PYTHON BINDS NAMES TO OBJECTS — NOT BOXES_. 🐍🧠**

In Python, variables aren’t containers that *hold* data — they’re **names** bound to **objects** in memory.
This notebook explores:
- The difference between *rebinding* and *in-place mutation*
- How `a = b` just binds another name to the same object
- Why `a += b` behaves differently for mutable vs immutable objects

## 📦 Import Standard Library

```python
import sys          # to inspect object identity
```

## 📝 Snippet 1 — Variables are names, not boxes

*Assignment in Python doesn’t copy data — it simply binds a name to an existing object.*

```python
a = [1, 2, 3]        # create a list object
b = a                # b is now another name pointing to the same list

print("a:", a)
print("b:", b)
print("a is b:", a is b)     # True → same object
print("id(a):", id(a))
print("id(b):", id(b))
```

## 🔁 Snippet 2 — Rebinding vs Mutation

*Rebinding assigns a new object to a name.
Mutation changes the contents of the existing object.*

```python
a = [1, 2, 3]
b = a               # both refer to same list
print("Before mutation:", a, b)

a.append(4)         # in-place mutation (list is mutable)
print("After mutation:", a, b)  # both reflect the change

a = [10, 20, 30]    # rebinding: a now points to a *new* list
print("After rebinding:", a, b) # b still refers to old list
```

## 🧩 Snippet 3 — Immutable objects behave differently

*Immutable types like ints, floats, strings, and tuples **cannot** be changed in place.
Every modification creates a new object.*

```python
x = 10
y = x               # both names point to same int object

print("Before:", x, y)
print("id(x):", id(x))
print("id(y):", id(y))

x += 5              # new int created (since ints are immutable)
print("\nAfter x += 5:", x, y)
print("id(x):", id(x))    # different id → new object
print("id(y):", id(y))    # unchanged
```

## ⚡ Snippet 4 — Mutable vs Immutable: Lists vs Tuples

*The `+=` operator behaves differently based on mutability.*

```python
# Mutable example
a = [1, 2]
b = a
a += [3, 4]     # list.__iadd__ mutates in place
print("Lists → a is b:", a is b)      # True

# Immutable example
x = (1, 2)
y = x
x += (3, 4)     # tuple.__add__ creates a new object
print("Tuples → x is y:", x is y)    # False
```

## 🧠 Snippet 5 — Visualizing Name Binding

*A name simply points to an object. Multiple names can point to the same one.*

```python
data = {"key": [1, 2, 3]}
alias = data

alias["key"].append(4)
print("Data:", data)
print("Alias:", alias)
print("Same object:", data is alias)
```

## ✅ One-liner Takeaway

**In Python, variables are *names bound to objects*, not boxes holding values — rebinding changes the name’s target, mutation changes the object itself.**
