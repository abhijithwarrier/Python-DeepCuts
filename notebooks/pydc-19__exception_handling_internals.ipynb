**PYTHON SCRIPT TO _UNDERSTAND HOW PYTHON HANDLES EXCEPTIONS INTERNALLY_. 🐍🧠**

Exceptions in Python are not just error messages — they are objects that travel through the call stack.

In this DeepCut, we explore:
- how exceptions propagate
- how stack unwinding works
- how try/except blocks affect execution
- the cost of exceptions vs normal flow

---

## 📦 Import Standard Library

```python
import traceback
import time
```

---

## 🧩 Snippet 1 — Exceptions are class instances

All exceptions derive from BaseException.

```python
try:
    raise ValueError("Something went wrong")
except ValueError as e:
    print(type(e))
    print(isinstance(e, Exception))
```

---

## 🧠 Snippet 2 — Exceptions bubble up the call stack

```python
def level_one():
    level_two()

def level_two():
    level_three()

def level_three():
    raise RuntimeError("Boom!")

try:
    level_one()
except RuntimeError as e:
    print("Caught:", e)
```

---

## 🔄 Snippet 3 — Python unwinds the stack during exceptions

Frames are popped until a matching handler is found.

```python
def test():
    try:
        1 / 0
    finally:
        print("Cleanup runs even during exception")

try:
    test()
except ZeroDivisionError:
    print("Handled outside")
```

---

## 🧠 Snippet 4 — Python records traceback information

```python
try:
    1 / 0
except Exception:
    traceback.print_exc()
```

---

## ⚠️ Snippet 5 — try blocks are cheap, exceptions are expensive

```python
def no_exception():
    for _ in range(1_000_000):
        try:
            x = 1 + 1
        except:
            pass

def with_exception():
    for _ in range(1000):
        try:
            raise ValueError
        except:
            pass

start = time.time()
no_exception()
print("No exception:", time.time() - start)

start = time.time()
with_exception()
print("With exception:", time.time() - start)
```

---

## 🧬 Snippet 6 — Python matches exception types hierarchically

```python
try:
    raise ValueError
except TypeError:
    print("TypeError")
except ValueError:
    print("ValueError caught")
```

---

## 🧠 Snippet 7 — You can define your own exception types

```python
class MyError(Exception):
    pass

try:
    raise MyError("Custom error")
except MyError as e:
    print(e)
```

---

## ✅ One-liner Takeaway

**Exceptions are objects that propagate through the call stack — triggering stack unwinding until a matching handler is found.**
