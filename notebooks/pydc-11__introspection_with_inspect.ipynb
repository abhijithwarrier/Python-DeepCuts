**PYTHON SCRIPT TO EXPLORE PYTHON OBJECTS, FUNCTIONS, AND THE CALL STACK AT RUNTIME USING inspect. 🐍🧠**

Python is a highly introspective language.
At runtime, you can inspect:
- function signatures
- parameters and annotations
- source code
- call stacks and frames
- module and class metadata

In this DeepCut, we explore how Python *looks at itself* using the `inspect` module —
a core building block behind debuggers, frameworks, and tooling.

---

## 📦 Import Standard Library

```python
import inspect
```

---

## 🧩 Snippet 1 — Inspect function parameters and return annotations

Frameworks use this to validate inputs, build CLIs, and generate APIs.

```python
def greet(name: str, age: int = 30) -> str:
    return f"Hello {name}, age {age}"

sig = inspect.signature(greet)
print(sig)

for name, param in sig.parameters.items():
    print(name, "→", param.annotation, param.default)
```

---

## 🧠 Snippet 2 — Retrieve the actual source code of a function

This works when source is available (not in compiled extensions).

```python
source = inspect.getsource(greet)
print(source)
```

---

## 🔍 Snippet 3 — Examine the current call stack at runtime

Useful for debuggers, logging frameworks, and tracing tools.

```python
def level_one():
    level_two()

def level_two():
    stack = inspect.stack()
    for frame in stack:
        print(frame.function)

level_one()
```

---

## 🧱 Snippet 4 — Access local variables and execution context

Frames expose execution state: locals, globals, and code objects.

```python
def compute():
    x = 10
    y = 20
    frame = inspect.currentframe()
    print("Locals:", frame.f_locals)

compute()
```

---

## 🧬 Snippet 5 — Identify what kind of object you're dealing with

Used heavily in plugins, ORMs, serializers, and decorators.

```python
class User:
    pass

def fn():
    pass

print("Is function:", inspect.isfunction(fn))
print("Is class:", inspect.isclass(User))
print("Is method:", inspect.ismethod(User.__init__))
```

## 🔄 Snippet 6 — Enumerate members of a module or class

This is how tools discover available APIs dynamically.

```python
members = inspect.getmembers(inspect)
for name, value in members[:10]:
    print(name, "→", type(value))
```

---

## 🧠 Snippet 7 — Using inspect to build smarter decorators

Inspect function metadata at runtime for logging or validation.

```python
def log_call(func):
    sig = inspect.signature(func)

    def wrapper(*args, **kwargs):
        bound = sig.bind(*args, **kwargs)
        print(f"Calling {func.__name__} with {bound.arguments}")
        return func(*args, **kwargs)

    return wrapper

@log_call
def add(a, b):
    return a + b

add(2, 3)
```
