Global variables work fine in simple programs.

But in concurrent applications:
- threads run simultaneously
- async tasks overlap
- requests execute independently

Python solves this problem using `contextvars`.

In this DeepCut, we explore:
- execution context
- ContextVar
- thread-local vs async-local state
- real-world framework usage

---

## 📦 Import Standard Library

```python
from contextvars import ContextVar
import asyncio
```

---

## 🧩 Global variables become dangerous in concurrent systems

```python
current_user = "alice"

def process():
    print(current_user)

process()
```

Works fine.

But in:

* web servers
* async applications
* concurrent tasks

global state quickly becomes unreliable.

---

## 🧠 ContextVar stores values per execution context

```python
current_user = ContextVar("current_user")

current_user.set("alice")

print(current_user.get())
```

---

## 🔄 Each execution context gets its own value

```python
current_user = ContextVar("current_user")

current_user.set("alice")
print(current_user.get())

current_user.set("bob")
print(current_user.get())
```

---

## 🧬 ContextVars work safely with async code

```python
current_user = ContextVar("current_user")

async def worker(name):
    current_user.set(name)

    await asyncio.sleep(1)

    print(current_user.get())

await asyncio.gather(
    worker("alice"),
    worker("bob")
)
```

---

## 🔍 Context variables support rollback using tokens

```python
user = ContextVar("user")

token = user.set("alice")

print(user.get())

user.reset(token)

try:
    print(user.get())
except LookupError:
    print("Value reset")
```

Used internally by:

This enables request-specific data without passing variables everywhere.

**ContextVar provides execution-specific state, making it safe to manage data across threads and async tasks without relying on fragile global variables.**
