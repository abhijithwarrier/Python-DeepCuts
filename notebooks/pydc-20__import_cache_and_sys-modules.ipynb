**PYTHON SCRIPT TO _UNDERSTAND HOW PYTHON CACHES MODULES AND WHY IMPORTS BEHAVE THE WAY THEY DO_. 🐍🧠**

Python does not re-import modules every time you use `import`.

Instead, it caches modules in a global dictionary called `sys.modules`.

In this DeepCut, we explore:
- how module caching works
- why imports are fast after the first time
- how reload behaves
- real-world pitfalls in large systems

---

## 📦 Import Standard Library

```python
import sys
import importlib
```

---

## 🧩 Snippet 1 — Python stores imported modules in sys.modules

```python
import math

print("math in cache:", "math" in sys.modules)
print("Module object:", sys.modules["math"])
```

---

## 🧠 Snippet 2 — Subsequent imports reuse cached module

```python
import math
import math

print(id(math), id(sys.modules["math"]))
```

---

## 🔄 Snippet 3 — Top-level code executes only on first import

```python
# Create a file: demo_module.py
# print("Module executing")

import demo_module
import demo_module  # No second print
```

---

## 🧠 Snippet 4 — importlib.reload forces re-execution

```python
importlib.reload(demo_module)
```

👉 But reload has caveats:

- existing references are NOT updated
- objects created earlier remain old

---

## ⚠️ Snippet 5 — Reload does not update existing references

```python
from demo_module import some_function

importlib.reload(demo_module)

# some_function still refers to old version
```

👉 This causes subtle bugs in:

- notebooks
- hot-reload systems
- long-running apps

---

## 🧬 Snippet 6 — Partially initialized modules in sys.modules

```python
# module A imports B
# module B imports A

# During import:
# Python inserts module into sys.modules BEFORE execution completes
```

👉 This prevents infinite recursion but can expose incomplete modules.

---

## 🔍 Snippet 7 — You can modify sys.modules (use carefully)

```python
sys.modules["fake_module"] = "Hello"

import fake_module
print(fake_module)
```

---

## 🧠 Snippet 8 — Real-world implications

Understanding sys.modules explains:
- stale code after reload
- unexpected shared state
- circular import bugs
- plugin system behavior

---

## ✅ One-liner Takeaway

**Python caches modules in sys.modules — imports reuse existing objects, and reload behavior must be handled carefully.**
