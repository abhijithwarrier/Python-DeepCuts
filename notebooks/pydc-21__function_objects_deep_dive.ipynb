**PYTHON SCRIPT TO _EXPLORE FUNCTION OBJECTS — THEIR INTERNAL ATTRIBUTES AND HOW PYTHON TREATS FUNCTIONS AS FIRST-CLASS OBJECTS_. 🐍🧠**

Functions in Python are not just executable blocks — they are full objects with metadata, state, and structure.

In this DeepCut, we explore:
- function internals (`__code__`, `__defaults__`, `__closure__`)
- how arguments and closures are stored
- how Python represents functions at runtime

---

## 📦 Import Standard Library

```python
import inspect
```

---

## 🧩 Snippet-1 -- Functions are first-class objects

```python
def greet():
    return "hello"

print(type(greet))
print(callable(greet))
```

---

## 🧠 Snippet-2 -- The __code__ object stores function internals

```python
def add(a, b):
    return a + b

code = add.__code__

print("Arg count:", code.co_argcount)
print("Variable names:", code.co_varnames)
print("Bytecode length:", len(code.co_code))
```

---

## 🔄 Snippet-3 -- Default values are stored in __defaults__

```python
def greet(name="Guest"):
    return name

print(greet.__defaults__)
```

---

## 🧬 Snippet-4 -- Closures store captured variables

```python
def outer(x):
    def inner():
        return x
    return inner

fn = outer(10)

print(fn.__closure__)
print(fn.__closure__[0].cell_contents)
```

---

## 🔍 Snippet-5 -- Functions carry metadata

```python
def sample():
    """Example function"""
    pass

print(sample.__name__)
print(sample.__doc__)
print(sample.__module__)
```

---

## 🧠 Snippet-6 -- Functions can be passed, stored, and returned

```python
def add(a, b):
    return a + b

def execute(fn, x, y):
    return fn(x, y)

print(execute(add, 2, 3))
```

---

## ✅ One-liner Takeaway

**Functions in Python are full objects — carrying code, state, defaults, and closures — enabling powerful patterns like decorators and higher-order functions.**
