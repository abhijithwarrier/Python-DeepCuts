**PYTHON SCRIPT TO _UNDERSTAND HOW ASYNC/AWAIT WORKS INTERNALLY IN PYTHON_. 🐍🧠**

`async` and `await` look like magic — but they are built on
very concrete runtime mechanics.

In this DeepCut, we explore:
- what `async def` actually creates
- how `await` pauses execution
- how coroutines differ from threads
- how the event loop schedules tasks
- why async scales for I/O but not CPU

---

## 📦 Import Standard Library

```python
import asyncio
import inspect
```

---

## 🧩 Snippet 1 — async functions return coroutine objects

Calling an async function does NOT execute it.
It returns a coroutine object.

```python
async def greet():
    return "Hello async"

coro = greet()
print(coro)
print("Is coroutine:", inspect.iscoroutine(coro))
```

---

## 🧠 Snippet 2 — await pauses execution, it does not block

`await` tells the event loop:
"I’m waiting — run something else."

```python
async def task():
    print("Start task")
    await asyncio.sleep(1)
    print("Resume task")

asyncio.run(task())
```

---

## 🔄 Snippet 3 — Coroutines are specialized generators

Async functions compile into state machines,
similar to generators.

```python
async def sample():
    await asyncio.sleep(0)

print(sample().__await__())
```

---

## 🧠 Snippet 4 — The event loop drives execution

The event loop:
- runs coroutines
- suspends them at await points
- resumes them when ready

```python
async def one():
    await asyncio.sleep(1)
    print("One done")

async def two():
    await asyncio.sleep(1)
    print("Two done")

async def main():
    await asyncio.gather(one(), two())

asyncio.run(main())
```

---

## 🧬 Snippet 5 — Async does not bypass the GIL

Async runs on:
- a single thread
- a single CPU core

It scales by overlapping I/O, not CPU work.

```python
async def cpu_bound():
    total = 0
    for i in range(10_000_000):
        total += i
    return total

asyncio.run(cpu_bound())
```

---

## 🔍 Snippet 6 — Tasks wrap coroutines for scheduling

Tasks are how the event loop tracks execution.

```python
async def work():
    await asyncio.sleep(1)
    return "done"

async def main():
    task = asyncio.create_task(work())
    print("Task created")
    result = await task
    print("Result:", result)

asyncio.run(main())
```

---

## 🧠 Snippet 7 — Async shines when waiting dominates execution

Async is ideal when:
- network calls
- database queries
- file I/O
- timers

The event loop stays busy while tasks wait.

---

## ✅ One-liner Takeaway

**Async/Await is cooperative multitasking powered by an event loop — it overlaps I/O efficiently but does not provide CPU parallelism.**
