**PYTHON SCRIPT TO *DYNAMICALLY LOAD, RELOAD, AND CONTROL MODULES AT RUNTIME USING importlib*. 🐍🧠**

Python’s `import` statement looks static — but under the hood, imports are **fully dynamic**.

The `importlib` module exposes Python’s import machinery, allowing you to:

- load modules by name at runtime
- import modules from file paths
- reload modules without restarting the interpreter
- build plugin systems and extensible architectures

This DeepCut explores how Python’s import system can be controlled programmatically.

---

## 📦 Import Standard Library

```python
import importlib
import sys
from pathlib import Path
```

---

## 🧩 Snippet 1 — Import modules dynamically using importlib.import_module

This is equivalent to using the `import` statement, but allows runtime control.

```python
module_name = "math"

math_module = importlib.import_module(module_name)
print(math_module.sqrt(16))
```

This pattern is used in:

- configuration-driven systems
- feature flags
- plugin loaders

---

## 🧠 Snippet 2 — Dynamically load submodules

Useful when module paths are constructed at runtime.

```python
json_decoder = importlib.import_module("json.decoder")
print(json_decoder.JSONDecoder)
```

---

## 🔍 Snippet 3 — Python caches imported modules in sys.modules

Once imported, a module is reused unless explicitly reloaded.

```python
print("math" in sys.modules)
print(sys.modules["math"])
```

This explains why imports are fast after the first load.

---

## 🔄 Snippet 4 — Reload modules using importlib.reload

Useful in:

- REPLs
- notebooks
- development servers
- hot-reload workflows

```python
importlib.reload(math_module)
```

⚠️ Reloading does **not** reset existing references — this is an important caveat.

---

## 🧱 Snippet 5 — Load modules from file paths dynamically

This bypasses sys.path and is useful for plugins.

```python
module_path = Path("example_plugin.py")

spec = importlib.util.spec_from_file_location("example_plugin", module_path)
plugin = importlib.util.module_from_spec(spec)
spec.loader.exec_module(plugin)

# Access objects from the dynamically loaded module
print(dir(plugin))
```

(This assumes example_plugin.py exists in the working directory.)

---

## 🧬 Snippet 6 — Simple plugin system using importlib

Frameworks use this pattern to load extensions dynamically.

```python
def load_plugin(name):
    module = importlib.import_module(name)
    if hasattr(module, "register"):
        module.register()
    else:
        print(f"{name} has no register()")

# Example usage:
# load_plugin("my_plugin")
```

This approach powers:

- plugin architectures
- command registries
- feature modules

## 🧠 Snippet 7 — importlib is Python’s import system

The `import` statement itself is implemented using importlib.

This module exposes:

- finders
- loaders
- specs
- execution hooks

**importlib gives you full control over Python’s import system — enabling dynamic loading, reloading, and plugin-based architectures at runtime.**
