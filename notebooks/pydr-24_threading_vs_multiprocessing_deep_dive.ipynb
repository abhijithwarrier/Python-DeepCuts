**PYTHON SCRIPT TO _UNDERSTAND THE DIFFERENCES BETWEEN THREADS AND PROCESSES IN PYTHON_. 🐍🧠**

Python provides multiple ways to achieve concurrency. The two most common approaches are:
- threading
- multiprocessing

In this DeepCut, we explore:
- memory sharing
- process isolation
- GIL impact
- inter-process communication (IPC)
- when to choose each approach

----

## 📦 Import Standard Library

```python
import threading
import multiprocessing
import os
```

---

## 🧩 Threads run inside the same process

Threads share:
- memory
- variables
- resources

```python
shared_data = []

def worker():
    shared_data.append("thread")

t = threading.Thread(target=worker)

t.start()
t.join()

print(shared_data)
```

----

## 🧠 Processes are isolated

Each process has its own memory space.

```python
data = []

def worker():
    data.append("process")
    print("Inside process:", data)

p = multiprocessing.Process(target=worker)

p.start()
p.join()

print("Main process:", data)
```

---

## 🔄 Threads compete for the GIL

Only one thread executes Python bytecode at a time.

```python
def cpu_task():
    total = 0
    for i in range(10_000_000):
        total += i

threads = [
    threading.Thread(target=cpu_task)
    for _ in range(2)
]

for t in threads:
    t.start()

for t in threads:
    t.join()
```

---

## 🧬 Separate processes have separate interpreters

Each process gets its own GIL.

```python
processes = [
    multiprocessing.Process(target=cpu_task)
    for _ in range(2)
]

for p in processes:
    p.start()

for p in processes:
    p.join()
```

---

## 🔍 Threads vs processes at the OS level

```python
print("Main PID:", os.getpid())

# Inside worker functions
print("Worker PID:", os.getpid())
```

Processes receive unique PIDs. Threads do not.

---

## ⚠️ Processes require IPC

Memory is isolated.
Communication must be explicit.

```python
def worker(queue):
    queue.put("hello")

queue = multiprocessing.Queue()

p = multiprocessing.Process(
    target=worker,
    args=(queue,)
)

p.start()
print(queue.get())
p.join()
```

---

## 🧠 Threading vs Multiprocessing

Use Threading:
- I/O-bound work
- network calls
- file operations

Use Multiprocessing:
- CPU-bound work
- heavy computation
- parallel execution

---

## ✅ One-liner Takeaway

**Threads share memory but are limited by the GIL; processes isolate memory and enable true CPU parallelism.**
