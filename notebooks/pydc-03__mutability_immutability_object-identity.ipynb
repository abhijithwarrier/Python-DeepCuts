**PYTHON SCRIPT TO _UNDERSTAND HOW PYTHON HANDLES MUTABLE & IMMUTABLE OBJECTS IN MEMORY_. 🐍🧠**

Everything in Python is an object, but not all objects behave the same way.
Some can be *mutated in place* (like lists and dicts), while others cannot (like tuples and strings).
This DeepCut explores:
- The difference between *mutable* and *immutable* objects
- How *object identity* and *value equality* differ
- How Python optimizes memory using *caching* and *interning*

## 📦 Import Standard Library

```python
import sys      # to inspect object IDs and reference counts
```

## 🧩 Snippet 1 — Mutable vs Immutable Objects

*Mutable objects can change in place — their identity stays the same.
Immutable objects cannot — every modification creates a new object.*

```python
# Mutable example
nums = [1, 2, 3]
print("Before mutation:", nums, "id:", id(nums))
nums.append(4)
print("After mutation:", nums, "id:", id(nums))     # same ID

# Immutable example
name = "AW"
print("\nBefore change:", name, "id:", id(name))
name += " Dev"
print("After change:", name, "id:", id(name))       # new ID
```

---

## 🧭 Snippet 2 — Identity vs Equality

*Two objects can have the same value but different identities.*

```python
a = [1, 2, 3]
b = [1, 2, 3]

print("a == b:", a == b)        # True → same value
print("a is b:", a is b)        # False → different objects
print("id(a):", id(a))
print("id(b):", id(b))
```

---

## 🧠 Snippet 3 — Object Caching and Interning

*Python optimizes memory by reusing certain immutable objects like small integers and strings.*

```python
# Small integers (-5 to 256) are cached and reused
x = 256
y = 256
print("x is y (256):", x is y)       # True → cached

a = 257
b = 257
print("a is b (257):", a is b)       # False → new objects

# String interning for small constants
s1 = "hello"
s2 = "hello"
print("\ns1 is s2:", s1 is s2)       # True → interned by Python

# Dynamic strings are not always interned
x = "".join(["py", "thon"])
y = "".join(["py", "thon"])
print("x is y:", x is y)             # Might be False
```

## 🔍 Snippet 4 — Shared References in Mutable Objects

*Mutating one reference affects all names bound to that object.*

```python
a = [1, 2, 3]
b = a
print("Before mutation:", a, b)
a.append(99)
print("After mutation:", a, b)     # both changed
print("Same object:", a is b)
```

---

## ⚡ Snippet 5 — Mutable Default Pitfall

*Using mutable defaults in functions can lead to shared-state bugs.*

```python
def add_item(item, container=[]):     # ⚠️ BAD: default is shared
    container.append(item)
    return container

print(add_item("A"))
print(add_item("B"))     # unexpected accumulation
print(add_item("C"))

# Correct approach
def add_item_fixed(item, container=None):
    if container is None:
        container = []
    container.append(item)
    return container

print("\nFixed version:")
print(add_item_fixed("A"))
print(add_item_fixed("B"))
```

## 🧬 Snippet 6 — Reference Counting

*Every object tracks how many names reference it.
Use `sys.getrefcount()` to see how many bindings exist.*

```python
x = [1, 2, 3]
print("Ref count for x:", sys.getrefcount(x))   # +1 because of function argument
y = x
print("Ref count after aliasing:", sys.getrefcount(x))
del y
print("After deleting y:", sys.getrefcount(x))
```

---

## ✅ One-liner Takeaway

**Mutable objects change in place; immutable ones don’t.
Python reuses immutable objects through caching and interning — a design that saves memory and avoids surprises.**
