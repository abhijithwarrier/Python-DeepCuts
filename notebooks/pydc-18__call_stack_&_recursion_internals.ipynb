**PYTHON SCRIPT TO _UNDERSTAND HOW PYTHON MANAGES FUNCTION CALLS, STACK FRAMES, AND RECURSION LIMITS_. 🐍🧠**

Every function call in Python adds a frame to the call stack. Recursive calls stack these frames until a limit is reached.

In this DeepCut, we explore:
- how the call stack grows
- how recursion works internally
- Python’s recursion limits
- what happens during stack overflow

---

## 📦 Import Standard Library

```python
import inspect
import sys
```

---

## 🧩 Snippet 1 — Function calls are tracked using a stack

Each function call creates a new frame and pushes it onto the stack.

```python
def level_one():
    level_two()

def level_two():
    level_three()

def level_three():
    stack = inspect.stack()
    for frame in stack:
        print(frame.function)

level_one()
```

---

## 🧠 Snippet 2 — Recursive calls stack frames repeatedly

Each recursive call adds a new frame.

```python
def recursive(n):
    print(f"Depth: {n}")
    if n == 0:
        return
    recursive(n - 1)

recursive(5)
```

---

## 🔄 Snippet 3 — Python limits recursion depth to prevent crashes

```python
print("Recursion limit:", sys.getrecursionlimit())
```

---

## ⚠️ Snippet 4 — Exceeding recursion limit raises RecursionError

```python
def overflow():
    return overflow()

try:
    overflow()
except RecursionError as e:
    print("RecursionError:", e)
```

---

## 🧬 Snippet 5 — Preventing segmentation faults

Without a recursion limit:
- Python could overflow the C stack
- leading to interpreter crash

The limit protects the runtime.

---

## 🧠 Snippet 6 — You can change recursion limits, but it’s risky

```python
sys.setrecursionlimit(2000)
print("New limit:", sys.getrecursionlimit())
```

---

## 🔍 Snippet 7 — Python does NOT optimize tail recursion

Each recursive call still creates a new frame.

```python
def tail_recursive(n):
    if n == 0:
        return 0
    return tail_recursive(n - 1)

tail_recursive(10)
```

---

## ✅ One-liner Takeaway

**Python uses a call stack of frames for execution — recursion builds this stack, and limits exist to prevent interpreter crashes.**
