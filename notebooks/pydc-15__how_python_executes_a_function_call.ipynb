**PYTHON SCRIPT TO _UNDERSTAND WHAT ACTUALLY HAPPENS WHEN A PYTHON FUNCTION IS CALLED_. 🐍🧠**

A simple line like:

    result = add(2, 3)

Triggers a surprisingly complex execution pipeline inside CPython.

In this DeepCut, we explore:
- how Python creates stack frames
- how arguments are bound
- how local scopes are managed
- how return values flow back
- how this differs for normal vs async functions

---

## 📦 Import Standard Library

```python
import inspect
import sys
```

---

## 🧩 Snippet 1 — Every function call creates a frame object

Frames represent execution state.

```python
def demo():
    frame = inspect.currentframe()
    print("Function:", frame.f_code.co_name)
    print("Locals:", frame.f_locals)

demo()
```

---

## 🧠 Snippet 2 — Arguments are bound before the function body runs

Python maps arguments to parameters eagerly.

```python
def add(a, b):
    frame = inspect.currentframe()
    print("Arguments:", frame.f_locals)
    return a + b

add(2, 3)
```

---

## 🔄 Snippet 3 — Each call has its own local namespace

Local variables live only for the duration of the call.

```python
def counter():
    x = 0
    x += 1
    return x

counter()
counter()
```

---

## 🧠 Snippet 4 — Return values are passed back via the call stack

```python
def inner():
    return "done"

def outer():
    return inner()

print(outer())
```

---

## 🧬 Snippet 5 — async calls return coroutine objects

Execution is deferred.

```python
async def async_fn():
    return "async result"

coro = async_fn()
print("Returned:", coro)
```

---

## 🔍 Snippet 6 — Python tracks nested calls using a stack

```python
def level_one():
    level_two()

def level_two():
    stack = inspect.stack()
    for frame in stack:
        print(frame.function)

level_one()
```

## ✅ One-liner Takeaway

**Every Python function call creates a frame, binds arguments, executes in isolation, and returns through the call stack — async simply defers that execution.**
