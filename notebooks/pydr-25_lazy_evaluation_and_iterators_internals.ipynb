**PYTHON SCRIPT TO UNDERSTAND HOW PYTHON ITERATORS, GENERATORS, AND LAZY EVALUATION WORK INTERNALLY. 🐍🧠**

Python avoids unnecessary computation using lazy evaluation.

Instead of producing all values immediately, iterators and generators
produce values only when requested.

In this DeepCut, we explore:
- iter() and next()
- the iterator protocol
- generators
- lazy pipelines
- memory-efficient execution

# 🧩 Snippet 1 — The Iterator Protocol
## 🧩 Every iterator follows the iterator protocol

An iterator must implement:
- __iter__()
- __next__()

```python
numbers = [1, 2, 3]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
```

👉 iter() creates an iterator.
👉 next() asks for the next value.

---

# 🧠 Snippet 2 — StopIteration Ends the Sequence
## 🧠 Iterators signal completion with StopIteration

```python
numbers = [1]

iterator = iter(numbers)

print(next(iterator))

try:
    print(next(iterator))
except StopIteration:
    print("Iterator exhausted")
```

---

# 🔄 Snippet 3 — What a for Loop Really Does
## 🔄 A for loop repeatedly calls next()

```python
items = ["a", "b", "c"]

iterator = iter(items)

while True:
    try:
        item = next(iterator)
        print(item)
    except StopIteration:
        break
```

---

# 🧬 Snippet 4 — Generators Are Lazy Iterators
## 🧬 Generators produce values on demand

```python
def count_to_three():
    yield 1
    yield 2
    yield 3

gen = count_to_three()

print(next(gen))
print(next(gen))
print(next(gen))
```

---

# 🔍 Snippet 5 — Lazy Evaluation Saves Memory
## 🔍 Generators avoid storing everything in memory

```python
squares = (x * x for x in range(1_000_000))

print(next(squares))
print(next(squares))
```

---

# ⚠️ Snippet 6 — Generators Are Single-Use
## ⚠️ Generators are consumed as they iterate

```python
gen = (x for x in range(3))

for item in gen:
    print(item)

print(list(gen))
```

---

# 🧠 Snippet 7 — Building Lazy Pipelines
## 🧠 Generators can be chained into pipelines

```python
numbers = range(10)

evens = (x for x in numbers if x % 2 == 0)

squares = (x * x for x in evens)

print(list(squares))
```

---

## ✅ One-liner Takeaway

**Iterators and generators power Python’s lazy evaluation model — producing values only when requested and enabling memory-efficient execution.**
